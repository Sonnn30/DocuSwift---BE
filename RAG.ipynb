```python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
def read_file(file_path: str):
    reader = PyPDFLoader(file_path)
    pages = reader.load()

    return pages
```

```python

file = read_file("Monthly Report Juni WIlson Prajnawira.pdf")

# print(file)

```

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=200,
    chunk_overlap=20
)

chunk = text_splitter.split_documents(file)
```

```python
print(chunk)
```

```python
embed_model = HuggingFaceEmbeddings(
    model_name="paraphrase-multilingual-MiniLM-L12-v2"
)
texts = [doc.page_content for doc in chunk]
embedding = embed_model.embed_documents(texts)
```

```python
print(embedding)
```
